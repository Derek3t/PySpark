### **Análise de Redes Complexas e Detecção de Fraudes em Grafos (GraphX / Relacional)**
**O Cenário Real**: Redes sociais (como **Meta** ou **LinkedIn**) precisam mapear bilhões de conexões entre usuários instantaneamente, e **bancos** precisam identificar redes de lavagem de dinheiro onde várias contas laranjas transferem pequenas quantias entre si para burlar o sistema.

**Como o Spark atua**: Através da extensão de grafos e junções relacionais massivas, o Spark modela entidades como Vértices (usuários, contas, empresas) e transações ou amizades como Arestas (Edges), aplicando algoritmos de propagação e análise de caminhos em larga escala.

**Por que é relevante**: Analisar grafos de relacionamento em bancos de dados relacionais tradicionais (SQL clássico) gera múltiplos j**oins pesados que travam o sistema**. Com o processamento distribuído do Spark, é possível mapear comunidades inteiras e calcular relevância de nós (como o algoritmo **PageRank**) em um cluster paralelo.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

def main():
    spark = (SparkSession.builder
             .appName("Modulo_3_GraphAnalysis")
             .master("local[*]")
             .getOrCreate())
    
    spark.sparkContext.setLogLevel("WARN")
    print("=========================================================")
    print("=== PILAR 3: ANALISE DE REDES E GRAFOS (CONCEITO GRAPHX) ==")
    print("=========================================================")

    sales_data = [
        ("GOOG", "Sam", 200.0),
        ("GOOG", "Charlie", 120.0),
        ("GOOG", "Frank", 340.0),
        ("MSFT", "Tina", 600.0),
        ("MSFT", "Amy", 124.0),
        ("MSFT", "Vanessa", 243.0),
        ("FB", "Carl", 870.0),
        ("FB", "Sarah", 350.0),
        ("APPL", "John", 250.0),
        ("APPL", "Linda", 130.0),
        ("APPL", "Mike", 750.0),
        ("APPL", "Chris", 350.0)
    ]
    df = spark.createDataFrame(sales_data, ["Company", "Person", "Sales"])

    # Definicao de Vertices (Pessoas)
    vertices = df.select(col("Person").alias("id")).distinct()
    print("\n--- Vertices (Entidades Unicas) ---")
    vertices.show(5)

    # Definicao de Arestas (Relacoes de Transacao entre Origem e Destino)
    edges = df.select(
        col("Person").alias("src"), 
        col("Company").alias("dst"), 
        col("Sales").alias("weight")
    )

    print("--- Arestas Filtradas (Conexoes de Transacoes de Alto Valor > 300) ---")
    edges.filter(col("weight") > 300).show()

    spark.stop()

if __name__ == "__main__":
    main()
```
